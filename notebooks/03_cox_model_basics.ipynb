```python
import pandas as pd
from lifelines import CoxPHFitter
```

```python
cox_data = pd.DataFrame({
    "age": [63, 55, 71, 49, 66, 58, 74, 52],
    "hpv_positive": [1, 0, 1, 1, 0, 1, 0, 1],
    "rfs_days": [800, 250, 1200, 430, 600, 310, 500, 900],
    "event": [0, 1, 0, 1, 1, 0, 1, 0]
})

cox_data
```

```python
cph = CoxPHFitter()

cph.fit(
    cox_data,
    duration_col="rfs_days",
    event_col="event"
)

cph.print_summary()
```

```python
cph.check_assumptions(cox_data, p_value_threshold=0.05)
```
