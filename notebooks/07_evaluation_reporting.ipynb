```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
results = pd.DataFrame({
    "centre": ["A", "B", "C"],
    "c_index": [0.200, 0.333, 0.200]
})

results
```

```python
summary = {
    "mean_c_index": results["c_index"].mean(),
    "std_c_index": results["c_index"].std(),
    "min_c_index": results["c_index"].min(),
    "max_c_index": results["c_index"].max()
}

summary
```

```python
results.plot(
    x="centre",
    y="c_index",
    kind="bar",
    legend=False
)

plt.ylabel("C-index")
plt.title("Held-out Centre Performance")
plt.ylim(0, 1)
plt.show()
```

```python
results.to_csv("../results/centre_wise_c_index.csv", index=False)

print("Saved centre-wise results.")
```

```python
summary_df = pd.DataFrame([summary])

summary_df.to_csv("../results/summary_metrics.csv", index=False)

print("Saved summary metrics.")
```
