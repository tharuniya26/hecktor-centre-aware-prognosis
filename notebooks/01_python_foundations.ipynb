```python
import pandas as pd

data = {
    "patient_id": ["P001", "P002", "P003"],
    "age": [63, 55, 71],
    "hpv": ["positive", "negative", "positive"]
}

patients = pd.DataFrame(data)

patients
```

```python
print(patients.shape)
print(patients.columns)
print(patients["age"].mean())
```

```python
older_patients = patients[patients["age"] > 60]

older_patients
```

```python
# 1. Filter patients older than 60
older_patients = patients[patients["age"] > 60]
print(older_patients)

# 2. Add survival-style columns
patients["rfs_days"] = [800, 250, 1200]
patients["event"] = [0, 1, 0]

print(patients)

# 3. Count observed events
print("Number of events:", patients["event"].sum())

# 4. Average age
print("Mean age:", patients["age"].mean())
```
