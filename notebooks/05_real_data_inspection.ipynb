```python
from pathlib import Path

data_dir = Path("../data/raw")

print("Raw data folder:", data_dir)
print("Exists:", data_dir.exists())
```

```python
images_dir = data_dir / "imagesTr"
labels_dir = data_dir / "labelsTr"

print("Images folder exists:", images_dir.exists())
print("Labels folder exists:", labels_dir.exists())
```

```python
image_files = list(images_dir.iterdir())
label_files = list(labels_dir.iterdir())

print("Number of files in imagesTr:", len(image_files))
print("Number of files in labelsTr:", len(label_files))
```
