```python
import pandas as pd
```

```python
data = pd.DataFrame({
    "patient_id": ["P001", "P002", "P003", "P004", "P005", "P006"],
    "centre": ["A", "A", "B", "B", "C", "C"],
    "age": [63, 55, 71, 49, 66, 58],
    "rfs_days": [800, 250, 1200, 430, 600, 310],
    "event": [0, 1, 0, 1, 1, 0]
})

data
```

```python
train_data = data[data["centre"] != "C"]
test_data = data[data["centre"] == "C"]

print("Training data:")
print(train_data)

print("\nTest data:")
print(test_data)
```

```python
def split_by_centre(data, test_centre):
    train_data = data[data["centre"] != test_centre]
    test_data = data[data["centre"] == test_centre]
    return train_data, test_data

train_data, test_data = split_by_centre(data, "B")

print("Training centres:")
print(train_data["centre"].unique())

print("Test centre:")
print(test_data["centre"].unique())
```

```python
for test_centre in data["centre"].unique():
    train_data, test_data = split_by_centre(data, test_centre)

    print("Held-out centre:", test_centre)
    print("Training centres:", train_data["centre"].unique())
    print("Test centres:", test_data["centre"].unique())
    print("---")
```
